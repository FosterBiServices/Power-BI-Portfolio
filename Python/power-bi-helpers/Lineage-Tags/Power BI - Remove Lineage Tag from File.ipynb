```python
import re
from pathlib import Path


import re
from pathlib import Path

# ---- CONFIG ----
input_file = Path(r"C:\Users\<USER>\<LOCATION>\ExitInterviews.txt")
output_file = Path(r"C:\Users\<USER>\<LOCATION>\performanceOutcomes_tmdl_noLineage.txt")

# Matches TMDL lineage properties like:
#   lineageTag: guid
#   lineageId: guid
#   sourceLineageTag: guid
LINEAGE_PATTERN = re.compile(
    r'''
    ^\s*                    # optional indentation
    (annotation\s+)?        # optional 'annotation '
    lineage(Tag|Id)?        # lineageTag or lineageId
    \s*:\s*                 # colon syntax
    [A-Fa-f0-9\-]+          # GUID
    \s*$
    ''',
    re.IGNORECASE | re.VERBOSE
)

removed_count = 0

with input_file.open("r", encoding="utf-8") as fin, \
     output_file.open("w", encoding="utf-8") as fout:

    for line in fin:
        if LINEAGE_PATTERN.match(line):
            removed_count += 1
            continue  # skip lineage line entirely
        fout.write(line)

print(f"✅ Removed {removed_count} lineage entries")
print(f"✅ Output written to: {output_file}")

```
