# TMDL to Text Converter

Convert Power BI Tabular Model Definition Language (TMDL) files into plain-text files that can be uploaded to Microsoft Copilot.

## Purpose

Some organizations restrict direct connections to Power BI skills, Model Context Protocol (MCP) servers, or other external integrations. In those environments, this script provides a file-based alternative for sharing TMDL content with Copilot.

The script converts TMDL files to `.txt` files. It does not change the model logic or interpret the TMDL. The output is intended to make the model definition available in a file format accepted by the user's Copilot environment.

> **Important:** Converting a file to text does not remove sensitive information. Review the generated files and follow your organization's data-handling, privacy, and Copilot usage policies before uploading them.

## Requirements

- Python 3
- Access to the TMDL files or folder to be converted
- Permission to process and upload the model content
- A Copilot environment that accepts plain-text file uploads

## How to Use

1. Export or locate the TMDL files for the Power BI semantic model.
2. Place the Python script in an accessible folder.
3. Run the script with Python.
4. Select or provide the TMDL files requested by the script.
5. Allow the script to create the corresponding `.txt` output files.
6. Review the generated text files for sensitive or restricted information.
7. Upload only the approved `.txt` files to Copilot.
8. Include a clear prompt explaining what you want Copilot to review, document, troubleshoot, or summarize.

Review this content before uploading it outside the approved company environment.

## Data Handling and Security

Before uploading generated files to Copilot:

- Confirm that your organization permits the model content to be used in Copilot.
- Check for server names, database names, paths, query text, internal terminology, or confidential business logic.
- Remove content that is not required for the task.
- Use the organization-approved Copilot account and tenant.
- Do not assume that changing the extension from `.tmdl` to `.txt` anonymizes or sanitizes the content.
- Retain the original TMDL files as the source of record.

## Limitations

- The conversion changes file accessibility, not the underlying sensitivity of the content.
- Copilot can only analyze the files provided and may not have the full semantic-model context.
- Missing files can lead to incomplete analysis.
- Large models may need to be divided into multiple uploads because file limits vary by Copilot environment and organizational configuration.
- The generated text files are reference copies and should not replace the original TMDL source files.

## Troubleshooting

### No output files are created

Confirm that the script can access the source location and that the selected path contains TMDL files. Also confirm that the output location is writable.

### Output files are empty

Verify that the source TMDL files contain text and were read using the expected character encoding.

### Some model objects are missing in Copilot's response

Confirm that all related text files were uploaded and tell Copilot to use the file names and paths as structural context. For larger models, ask Copilot to identify which supplied files it used in its response.

### Copilot cannot process all files together

Split the files into logical groups, such as model metadata, relationships, measures, and table definitions. Provide the same model context in each prompt and avoid separating files that depend on one another.

````python
from pathlib import Path
import shutil

# =============================================================================
# Configuration
# =============================================================================
_ModelName = input("Enter semantic model name: ").strip()

#SOURCE_FOLDER = Path(fr"C:\Users\<user>\<LOCATION>\{_ModelName}\{_ModelName}.SemanticModel\definition")
#SOURCE_FOLDER = Path(fr"C:\Users\<user>\<LOCATION>\{_ModelName}\{_ModelName}.SemanticModel\definition\tables")
SOURCE_FOLDER = Path(fr"C:\Users\<user>\<LOCATION>\{_ModelName}.SemanticModel\definition")

OUTPUT_FOLDER = Path(fr"C:\Users\<user>\<LOCATION>\{_ModelName}_TMDL_Text_Files")

# =============================================================================
# Convert TMDL Files to TXT
# =============================================================================
def convert_tmdl_to_txt(source_folder: Path, output_folder: Path) -> None:
    """
    Converts all .tmdl files under source_folder to .txt files while
    preserving the folder structure.
    """

    output_folder.mkdir(parents=True, exist_ok=True)

    tmdl_files = list(source_folder.rglob("*.tmdl"))

    if not tmdl_files:
        print("No TMDL files found.")
        return

    converted_count = 0

    for tmdl_file in tmdl_files:
        try:
            relative_path = tmdl_file.relative_to(source_folder)

            destination_folder = output_folder / relative_path.parent
            destination_folder.mkdir(parents=True, exist_ok=True)

            destination_file = destination_folder / f"{tmdl_file.stem}.txt"

            content = tmdl_file.read_text(
                encoding="utf-8",
                errors="replace"
            )

            destination_file.write_text(
                content,
                encoding="utf-8"
            )

            converted_count += 1
            print(f"Converted: {tmdl_file}")

        except Exception as ex:
            print(f"Failed: {tmdl_file}")
            print(f"Error : {ex}")

    print("-" * 80)
    print(f"Completed. {converted_count} file(s) converted.")
    print(f"Output Folder: {output_folder}")

# =============================================================================
# Main
# =============================================================================
if __name__ == "__main__":
    convert_tmdl_to_txt(
        source_folder=SOURCE_FOLDER,
        output_folder=OUTPUT_FOLDER
    )
````
