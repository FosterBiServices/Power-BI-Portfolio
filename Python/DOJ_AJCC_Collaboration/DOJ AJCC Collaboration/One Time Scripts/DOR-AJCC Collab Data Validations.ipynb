```python
import pandas as pd
import numpy as np
from pathlib import Path

# Update this path if needed
DATA_DIR = Path( r"C:\Users\StevenFoster\APM US\Data and Insights - Documents\Power BI Source Files\DOR-AJCC Collab")

regions = pd.read_excel(DATA_DIR / 'Regions.xlsx')
participants = pd.read_excel(DATA_DIR / '(All) DOR-AJCC Collab_GrantParticipants.xlsx')
services = pd.read_excel(DATA_DIR / '(All) DOR-AJCC Collab_SevicesProvided.xlsx')

for name, df in {
    "regions": regions,
    "participants": participants, 
    "services": services
}.items():

    profile_rows = []

    for column in df.columns:

        series = df[column]

        row_count = len(series)
        null_count = series.isna().sum()
        non_null_count = row_count - null_count
        distinct_count = series.nunique(dropna=True)

        # Blank count for text/object columns
        if series.dtype == "object":
            blank_count = (
                series.fillna("")
                .astype(str)
                .str.strip()
                .eq("")
                .sum()
            )
        else:
            blank_count = 0

        profile_rows.append(
            {
                "Column": column,
                "DataType": str(series.dtype),
                "RowCount": row_count,
                # "NonNullCount": non_null_count,
                # "NullCount": null_count,
                # "NullPct": round((null_count / row_count) * 100, 2),
                # "BlankCount": blank_count,
                # "NullOrBlankCount": null_count + blank_count,
                "DistinctCount": distinct_count,
                # "DistinctPct": round(
                    # (distinct_count / max(non_null_count, 1)) * 100,
                    # 2
                # ),
                "UniqueValueCount": (
                    series.value_counts(dropna=True)
                    .eq(1)
                    .sum()
                ),
                "DuplicateCount": max(
                    non_null_count - distinct_count,
                    0
                ),
                "IsFullyUnique": distinct_count == non_null_count,
                "HasDuplicates": distinct_count < non_null_count,
                "IsNullable": null_count > 0,
            }
        )

    profile_df = pd.DataFrame(profile_rows)

    print("\n" + "=" * 80)
    print(name.upper())
    print("Shape:", df.shape)

    display(profile_df.sort_values("Column"))
```
